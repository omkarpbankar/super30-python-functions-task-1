# Custom Functions and Parameters

This notebook practices reusable programming with function definitions, calls, parameters, arguments, return values, scope, and default arguments. The problem-solving logic for each exercise is inside its function.

## Function vocabulary

- A **function definition** uses `def` to describe reusable behavior.
- A **parameter** is a name in the function definition, such as `a` and `b` in `add(a, b)`.
- A **function call** runs the function, such as `add(2, 3)`.
- An **argument** is a value supplied in a call. Arguments can be **positional** (`add(2, 3)`) or **keyword** (`add(a=2, b=3)`).
- `return` sends a result back to the caller so it can be stored, printed, or used in another expression.
- A **default argument** supplies a value when the caller omits that argument, such as `discount=10`.

```python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b


def subtract(a, b):
    """Return the difference between two numbers."""
    return a - b


def multiply(a, b):
    """Return the product of two numbers."""
    return a * b


def divide(a, b):
    """Return the quotient of two numbers and reject division by zero."""
    if b == 0:
        raise ValueError("The divisor cannot be zero.")
    return a / b


print(add(10, 5))
print(subtract(10, 5))
print(multiply(10, 5))
print(divide(10, 5))
```

```python
def even_or_odd(number):
    """Return whether a number is even or odd."""
    return "Even" if number % 2 == 0 else "Odd"


def largest_of_three(first, second, third):
    """Return the largest of three numbers without using max()."""
    largest = first
    if second > largest:
        largest = second
    if third > largest:
        largest = third
    return largest


def factorial(number):
    """Return the factorial of a non-negative integer."""
    if number < 0 or number != int(number):
        raise ValueError("Factorial requires a non-negative integer.")
    result = 1
    for value in range(2, int(number) + 1):
        result *= value
    return result


def is_prime(number):
    """Return True when number is a prime number."""
    if number < 2 or number != int(number):
        return False
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            return False
    return True


print(even_or_odd(7))
print(largest_of_three(12, 45, 23))
print(factorial(5))
print(is_prime(29))
```

```python
def calculate_discount(price, discount=10):
    """Return the price after applying a percentage discount."""
    if price < 0 or discount < 0 or discount > 100:
        raise ValueError("Price must be non-negative and discount must be 0 to 100.")
    return price - (price * discount / 100)


# The default argument applies when discount is omitted.
print(calculate_discount(100))
print(calculate_discount(100, 25))
```

```python
def list_sum(numbers):
    """Return the sum of a list without using sum()."""
    total = 0
    for number in numbers:
        total += number
    return total


def count_vowels(text):
    """Return the number of vowels in a string."""
    vowels = "aeiou"
    count = 0
    for character in text.lower():
        if character in vowels:
            count += 1
    return count


def is_palindrome(text):
    """Return True when text reads the same forwards and backwards."""
    normalized = "".join(character.lower() for character in text if character.isalnum())
    return normalized == normalized[::-1]


print(list_sum([4, 7, 9]))
print(count_vowels("Reusable Functions"))
print(is_palindrome("Never odd or even"))
```

```python
def student_profile(name, age, course):
    """Return a formatted profile for a student."""
    return f"Student: {name} | Age: {age} | Course: {course}"


# Positional arguments match parameters by their order.
positional_profile = student_profile("Asha", 20, "Python")

# Keyword arguments match parameters by their names and can be reordered.
keyword_profile = student_profile(course="Data Science", name="Ravi", age=21)

print(positional_profile)
print(keyword_profile)
```

## `print()` versus `return`

`print()` displays a value immediately but returns `None`. `return` gives the value back to the caller, so it can be stored or combined with other functions.

```python
def show_total(first, second):
    """Print a total without returning it."""
    print(first + second)


def get_total(first, second):
    """Return a total so the caller can reuse it."""
    return first + second


printed_result = show_total(2, 3)
returned_result = get_total(2, 3)
reused_result = multiply(returned_result, 10)

print("Value received from show_total:", printed_result)
print("Value received from get_total:", returned_result)
print("Returned value reused in another function:", reused_result)
```

## Scope

Variables created inside a function are local to that function. The `result` variable below can be used inside `scope_demo`, but it is not available outside the function.

```python
def scope_demo(number):
    """Return a local calculation to demonstrate function scope."""
    result = number * 2
    return result


scope_value = scope_demo(6)
print("The returned local value is:", scope_value)
# `result` cannot be referenced here because it only exists inside scope_demo.
```

```python
# Small checks confirm the functions return the expected values.
assert add(2, 3) == 5
assert subtract(8, 3) == 5
assert multiply(4, 3) == 12
assert divide(9, 3) == 3
assert even_or_odd(8) == "Even"
assert largest_of_three(8, 12, 4) == 12
assert factorial(0) == 1 and factorial(5) == 120
assert is_prime(29) and not is_prime(1)
assert calculate_discount(200) == 180
assert list_sum([1, 2, 3]) == 6
assert count_vowels("Education") == 5
assert is_palindrome("Level")
assert printed_result is None and returned_result == 5
print("All checks passed.")
```
